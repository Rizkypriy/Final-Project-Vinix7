## Data Preparation

```python
df_clean = df.copy()
print("Ukuran Data Mentah:", df.shape)
print("Ukuran Data Preparation:", df_clean.shape)
```

```python
df_clean['EMAIL INFO PROGRAM'] = (
    df_clean['EMAIL INFO PROGRAM']
    .astype('boolean')
    .fillna(False)
)
```

```python
for column in ['Kode Referensi', 'Tempat Magang',
               'Perguruan Tinggi', 'Program Studi', 'Dream Job']:

    print(f"\n{column}")
    print(df_clean[df_clean[column].astype(str).str.strip() == '-'][column].head())
```

```python
df_clean['Timestamp'] = pd.to_datetime(
    df_clean['Timestamp'],
    errors='coerce'
)
```

```python
missing_after = pd.DataFrame({
    'Missing Value': df_clean.isnull().sum(),
    'Missing (%)': (
        df_clean.isnull().sum() / len(df_clean) * 100
    ).round(2)
})

missing_after[missing_after['Missing Value'] > 0].sort_values(
    by='Missing Value',
    ascending=False
)
```

```python
columns_placeholder = [
    'Kode Referensi',
    'Tempat Magang',
    'Perguruan Tinggi',
    'Program Studi',
    'Dream Job'
]

for column in columns_placeholder:
    df_clean[column] = df_clean[column].replace('-', pd.NA)

print("berhasil diubah")
```

```python
for column in df_clean.select_dtypes(include='object').columns:
    df_clean[column] = df_clean[column].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )
```

```python
columns_to_check = [
    'Perguruan Tinggi',
    'Program Studi',
    'Dream Job'
]

for column in columns_to_check:
    print(f'\n--- {column} ---')

    temp = df_clean[column].dropna()

    normalized = temp.str.casefold()

    variants = (
        pd.DataFrame({
            'Original': temp,
            'Normalized': normalized
        })
        .groupby('Normalized')['Original']
        .agg(lambda x: sorted(set(x)))
    )

    variants = variants[
        variants.apply(len) > 1
    ]

    display(variants.head(20))
```

```python
import unicodedata

def inspect_text(value):
    if isinstance(value, str):
        return [
            (char, f'U+{ord(char):04X}', unicodedata.name(char, 'UNKNOWN'))
            for char in value
            if not char.isprintable() or ord(char) > 127
        ]

columns_to_check = [
    'Perguruan Tinggi',
    'Program Studi',
    'Dream Job'
]

for column in columns_to_check:
    print(f'\n--- {column} ---')

    for value in df_clean[column].dropna().unique():
        unusual = inspect_text(value)

        if unusual:
            print(repr(value), unusual)
```

```python
# normalisasi uicode
import unicodedata
import re

columns_to_normalize = [
    'Perguruan Tinggi',
    'Program Studi',
    'Dream Job'
]

for column in columns_to_normalize:
    df_clean[column] = df_clean[column].apply(
        lambda x: re.sub(
            r'\s+',
            ' ',
            unicodedata.normalize('NFKC', x)
        ).strip() if isinstance(x, str) else x
    )
```

```python
for column in columns_to_normalize:
    print(f'\n--- {column} ---')

    print(
        'Jumlah kategori:',
        df_clean[column].nunique()
    )

    print(
        'Contoh data:',
        df_clean[column].dropna().head(10).tolist()
    )
```

```python
# periksa perguruan tinggi
column = 'Perguruan Tinggi'

temp = df_clean[column].dropna()

variants = (
    temp.groupby(temp.str.casefold())
    .agg(
        Variasi_Nama=lambda x: sorted(x.unique()),
        Jumlah_Baris='size'
    )
    .reset_index(drop=True)
)

variants['Jumlah_Variasi'] = variants['Variasi_Nama'].apply(len)

variants = variants[
    variants['Jumlah_Variasi'] > 1
].sort_values(
    'Jumlah_Baris',
    ascending=False
)

display(variants)
```

```python
column = 'Perguruan Tinggi'

temp = df_clean[column].dropna()

mapping = {}

for key, group in temp.groupby(temp.str.casefold()):
    canonical = group.value_counts().index[0]

    for value in group.unique():
        mapping[value] = canonical

# Tampilkan mapping yang memiliki variasi
mapping_review = pd.DataFrame(
    mapping.items(),
    columns=['Original', 'Standard']
)

mapping_review = mapping_review[
    mapping_review['Original'] != mapping_review['Standard']
]

display(mapping_review.head(50))

print('Jumlah nama yang akan distandardisasi:', len(mapping_review))
```

```python
column = 'Perguruan Tinggi'

# Simpan jumlah kategori sebelum standardisasi
before = df_clean[column].nunique()

# Terapkan mapping
df_clean[column] = df_clean[column].replace(mapping)

# Hitung jumlah kategori setelah standardisasi
after = df_clean[column].nunique()

print('Kategori sebelum:', before)
print('Kategori sesudah:', after)
print('Kategori berkurang:', before - after)
```

```python
temp = df_clean['Perguruan Tinggi'].dropna()

remaining_variants = (
    temp.groupby(temp.str.casefold())
    .agg(lambda x: sorted(x.unique()))
)

remaining_variants = remaining_variants[
    remaining_variants.apply(len) > 1
]

print('Sisa kelompok variasi kapitalisasi:', len(remaining_variants))

display(remaining_variants)
```

```python
# Menghitung frekuensi setiap nama perguruan tinggi
pt_frequency = (
    df_clean['Perguruan Tinggi']
    .value_counts(dropna=False)
    .reset_index()
)

pt_frequency.columns = [
    'Perguruan Tinggi',
    'Frekuensi'
]

# Mengurutkan berdasarkan frekuensi terkecil
pt_frequency = pt_frequency.sort_values(
    by=['Frekuensi', 'Perguruan Tinggi'],
    ascending=[True, True]
).reset_index(drop=True)

# Ringkasan hasil
print("Jumlah baris:", len(df_clean))
print("Jumlah kategori unik:", df_clean['Perguruan Tinggi'].nunique())
print("Jumlah missing value:", df_clean['Perguruan Tinggi'].isna().sum())

display(pt_frequency)
```

```python
import re

# Mengambil nilai yang tidak missing
pt_values = pt_frequency[
    pt_frequency['Perguruan Tinggi'].notna()
].copy()

# Deteksi kandidat berdasarkan pola
pt_values['Perlu Review'] = pt_values.apply(
    lambda row: (
        row['Frekuensi'] <= 3
        or bool(re.fullmatch(
            r'[A-Z][A-Z0-9.&()/-]{1,9}',
            str(row['Perguruan Tinggi']).strip()
        ))
        or str(row['Perguruan Tinggi']).strip().lower() in [
            'institut',
            'universitas',
            'sekolah tinggi',
            'politeknik',
            'stie',
            'stmik',
            'uin',
            'iain',
            'stikes',
            'poltekkes',
            'fresh graduate',
            'test123'
        ]
        or str(row['Perguruan Tinggi']).strip().lower() in [
            '0', 'a', 'dd', 'i', '.'
        ]
    ),
    axis=1
)

# Menampilkan kandidat pemeriksaan
pt_review = pt_values[
    pt_values['Perlu Review']
].drop(columns='Perlu Review')

print("Jumlah kandidat yang perlu diperiksa:", len(pt_review))

display(pt_review)
```

```python
def standardize_jenjang(value):
    if not isinstance(value, str):
        return value

    # Normalisasi S1, S-1, S1-, dan S1 -
    value = re.sub(
        r'^\s*S\s*(?:-\s*1|1)\s*[- ]*\s*',
        'S1 ',
        value,
        flags=re.IGNORECASE
    )

    # Normalisasi D-3 dan D-4
    value = re.sub(
        r'^\s*D\s*-\s*([34])\s*',
        r'D\1 ',
        value,
        flags=re.IGNORECASE
    )

    return value.strip()


df_clean['Program Studi'] = (
    df_clean['Program Studi'].apply(standardize_jenjang)
)
```

```python
import re

def normalize_uiux(value):
    if not isinstance(value, str):
        return value

    # Standardisasi variasi UI/UX Design dan UI/UX Designer
    value = re.sub(
        r'UI\s*[/\\]\s*UX\s*(?:DESIGNER|DESAIGNER|DESAINER|DESIGHER|DESIGN)?(?!\s*DEVELOPER)',
        'UI/UX Designer',
        value,
        flags=re.IGNORECASE
    )

    # Koreksi typo khusus UI/UxXDesigner
    value = re.sub(
        r'UI\s*/\s*UXX\s*DESIGNER',
        'UI/UX Designer',
        value,
        flags=re.IGNORECASE
    )

    # Variasi tanpa tanda slash
    value = re.sub(
        r'UI\s+UX\s+DESIGN(?:ER)?',
        'UI/UX Designer',
        value,
        flags=re.IGNORECASE
    )

    return value

df_clean['Dream Job'] = (
    df_clean['Dream Job'].apply(normalize_uiux)
)
```

```python
from difflib import SequenceMatcher
from itertools import combinations

pt_values = (
    df_clean['Perguruan Tinggi']
    .dropna()
    .value_counts()
    .index
    .tolist()
)

pt_candidates = []

for a, b in combinations(pt_values, 2):
    similarity = SequenceMatcher(
        None,
        a.lower(),
        b.lower()
    ).ratio()

    if similarity >= 0.80:
        pt_candidates.append({
            'Nama 1': a,
            'Nama 2': b,
            'Kemiripan (%)': round(similarity * 100, 2)
        })

pt_candidates = pd.DataFrame(pt_candidates)

pt_candidates = pt_candidates.sort_values(
    'Kemiripan (%)',
    ascending=False
).reset_index(drop=True)

print("Jumlah pasangan kandidat:", len(pt_candidates))
display(pt_candidates)
```

```python
# Kelompokkan kandidat berdasarkan tingkat kemiripan
pt_review = pt_candidates.copy()

pt_review['Kategori Kandidat'] = pd.cut(
    pt_review['Kemiripan (%)'],
    bins=[79, 85, 90, 95, 100],
    labels=[
        '80–85%',
        '85–90%',
        '90–95%',
        '95–100%'
    ],
    include_lowest=True
)

# Tampilkan kandidat dengan kemiripan minimal 95%
pt_review_prioritas = (
    pt_review[pt_review['Kemiripan (%)'] >= 95]
    .sort_values('Kemiripan (%)', ascending=False)
    .reset_index(drop=True)
)

print('Jumlah kandidat prioritas:', len(pt_review_prioritas))

display(pt_review_prioritas)
```

```python
# 1. Hitung frekuensi setiap nama perguruan tinggi
pt_freq = (
    df_clean['Perguruan Tinggi']
    .dropna()
    .value_counts()
    .reset_index()
)

pt_freq.columns = [
    'Nama Perguruan Tinggi',
    'Frekuensi'
]

# 2. Identifikasi karakteristik nama
pt_freq['Panjang Nama'] = (
    pt_freq['Nama Perguruan Tinggi'].str.len()
)

pt_freq['Jumlah Kata'] = (
    pt_freq['Nama Perguruan Tinggi'].str.split().str.len()
)

# 3. Buat fungsi identifikasi nama yang perlu diperiksa
def cek_nama_pt(nama):
    alasan = []

    if len(nama) <= 20:
        alasan.append('Nama terlalu pendek')

    if nama.isupper():
        alasan.append('Seluruh huruf kapital')

    if any(char.isdigit() for char in nama):
        alasan.append('Mengandung angka')

    if any(char in nama for char in ['.', '/', '_', '@']):
        alasan.append('Mengandung karakter khusus')

    if len(nama.split()) == 1:
        alasan.append('Hanya satu kata')

    return ', '.join(alasan)

# 4. Terapkan identifikasi
pt_freq['Indikasi Masalah'] = (
    pt_freq['Nama Perguruan Tinggi'].apply(cek_nama_pt)
)

# 5. Ambil nama yang perlu diperiksa
pt_review_unclear = (
    pt_freq[
        pt_freq['Indikasi Masalah'] != ''
    ]
    .sort_values(
        ['Panjang Nama', 'Frekuensi'],
        ascending=[True, False]
    )
    .reset_index(drop=True)
)

# 6. Tampilkan hasil
print('Total nama perguruan tinggi unik:', len(pt_freq))
print('Jumlah kandidat perlu diperiksa:', len(pt_review_unclear))

display(pt_review_unclear)
```
